#Matrix Operations

```python
mat_1 = np.array([[1,2],[3,4]])
mat_2 = np.array([[5,6],[7,8]])
```

```python
print(mat_1)
print(mat_2)
```

Matrix Addition

```python
sum = mat_1 + mat_2
print(sum)
```

```python
#Another Way
sum = np.add(mat_1,mat_2)
print(sum)
```

Matric Subtraction

```python
diff = mat_1 - mat_2
print(diff)
```

```python
#or


diff = np.subtract(mat_1,mat_2)
print(diff)
```

Multiplying a Matrix by a Scalar

```python
mat = np.random.randint(20, size=(3,3))
print(mat)
```

```python
x = 2
mul = mat * x
print(mul)
```

```python
#or

mul = np.multiply(mat,x)
print(mul)
```

Multiplying Two Matrices

```python
matrix_1 = np.random.randint(10,size=(2,3))
matrix_2 = np.random.randint(10,size=(3,2))

print(matrix_1)
```

```python
print(matrix_2)
```

```python
multiplication = np.dot(matrix_1,matrix_2)
print(multiplication)
```
