```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set()
```

Plotting a Vector

```python
plt.quiver(0,0,4,5)
plt.show()
```

```python
plt.quiver(0,0,4,5,scale_units='xy', angles='xy', scale=1)
plt.xlim(-8,8)
plt.ylim(-8,8)
plt.show()
```

  Changing the Color

```python
plt.quiver(0,0,4,5,scale_units='xy', angles='xy', scale=1, color='r')
plt.xlim(-8,8)
plt.ylim(-8,8)
plt.show()
```

Plotting Multiple Vectors

```python
plt.quiver(0,0,3,4, scale_units='xy', angles='xy', scale=1)
plt.quiver(0,0,-1,-2, scale_units='xy', angles='xy', scale=1,color='r')
plt.xlim(-10,10)
plt.ylim(-10,10)
plt.show()
```

Addition of Two Vectors

```python
vector_1 = np.asarray([3,5])
vector_2 = np.asarray([2,-5])

sum = vector_1 + vector_2
print(sum)

```

```python
#Plotting two vectors
plt.quiver(0,0,3,5, scale_units='xy', angles='xy', scale=1)
plt.quiver(0,0,2,-5, scale_units='xy', angles='xy', scale=1)
plt.quiver(0,0,5,0, scale_units='xy', angles='xy', scale=1, color='r')

plt.xlim(-8,8)
plt.ylim(-8,8)
plt.show()
```

Subtraction of two vectors

```python
vector_1 = np.asarray([3,5])
vector_2 = np.asarray([2,-5])

difference = vector_1 - vector_2
print(difference)
```

```python
#Plotting the difference
plt.quiver(0,0,3,5,scale_units='xy', angles='xy', scale=1)
plt.quiver(0,0,2,-5,scale_units='xy', angles='xy', scale=1)

plt.quiver(0,0,1,10, scale_units='xy', angles='xy', scale=1, color='r')

plt.xlim(-10,10)
plt.ylim(-10,10)
plt.show()
```

Multiplying a vector by a scalar

```python
vector_1 = np.asarray([2,4])
vector_2 = 3*vector_1

print(vector_2)
```

```python
#Plotting the vector
plt.quiver(0,0,2,4,scale_units='xy', angles='xy', scale=1)
plt.quiver(0,0,6,12,scale_units='xy', angles='xy', scale=1, color='r')

plt.xlim(-15,15)
plt.ylim(-15,15)

plt.show()
```

#Complex Operations

Dot Product

```python
vector_1 = np.asarray([3,4])
vector_2 = np.asarray([2,-5])

dot = np.dot(vector_1, vector_2)
print(dot)
```

Cross Product

```python
a = np.asarray([3,4,5])
b = np.asarray([6,7,8])

cross = np.cross(a,b)
print(cross)
```

Projection of Vector

```python
x = np.asarray([23,45,62])
y = np.asarray([45,82,67])

magnitude_y = np.sqrt(np.sum(y**2))

projection = (np.dot(x,y)/magnitude_y**2)*y
print(projection)
```
